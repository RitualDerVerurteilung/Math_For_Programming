# Теория

Суть задания: выбрать корень 14-й степени из единицы по заданным условиям, подставить его в матрицу $A + \lambda I$ и найти определитель этой матрицы.

### 1. Выбор числа λ
Корни n-й степени из единицы задаются формулой Муавра:
$$ \lambda_k = \cos\frac{2\pi k}{n} + i\sin\frac{2\pi k}{n}, \quad k = 0, 1, \dots, n-1 $$

При n = 14 получаем $\lambda_k = \cos\frac{\pi k}{7} + i\sin\frac{\pi k}{7}$, где $k = 0, 1, \dots, 13$.

Условие $\operatorname{Im}\lambda > 0$ означает $\sin\frac{\pi k}{7} > 0$, то есть $0 < \frac{\pi k}{7} < \pi$. Отсюда $k = 1, 2, \dots, 6$.

Среди этих корней нужен тот, у которого действительная часть максимальна. Косинус убывает на промежутке $(0, \pi)$, поэтому максимум достигается при наименьшем k, то есть при $k = 1$:
$$ \lambda = \cos\frac{\pi}{7} + i\sin\frac{\pi}{7} $$

### 2. Матрица A + λI
Единичная матрица I умножается на число λ и складывается с матрицей A. Обе матрицы имеют размер 2 * 2, значит сложение корректно:
$$ A + \lambda I = \begin{pmatrix} 7 & 5 \\ 6 & 2 \end{pmatrix} + \begin{pmatrix} \lambda & 0 \\ 0 & \lambda \end{pmatrix} = \begin{pmatrix} 7 + \lambda & 5 \\ 6 & 2 + \lambda \end{pmatrix} $$

### 3. Определитель
Определитель матрицы размера 2 * 2 вычисляется по формуле:
$$ \det\begin{pmatrix} a & b \\ c & d \end{pmatrix} = ad - bc $$

Подставляем элементы:
$$ \det(A + \lambda I) = (7 + \lambda)(2 + \lambda) - 5 \cdot 6 = \lambda^2 + 9\lambda - 16 $$

### 4. Действительная и мнимая части
По формуле Муавра $\lambda^2 = \cos\frac{2\pi}{7} + i\sin\frac{2\pi}{7}$. Тогда:
$$ \operatorname{Re} = \cos\frac{2\pi}{7} + 9\cos\frac{\pi}{7} - 16, \quad \operatorname{Im} = \sin\frac{2\pi}{7} + 9\sin\frac{\pi}{7} $$

```python
import numpy as np

k = np.arange(14)
roots = np.exp(2j * np.pi * k / 14)

positive_imag = roots[roots.imag > 1e-12]
lam = positive_imag[np.argmax(positive_imag.real)]

I = np.eye(2)
A = np.array([
    [7, 5],
    [6, 2]
])

det = np.linalg.det(A + lam * I)

print(f"lambda = {lam.real:.6f} + {lam.imag:.6f}i")
print(f"Определитель: {det.real:.6f} + {det.imag:.6f}i")
print(f"1) Действительная часть: {det.real:.6f}")
print(f"2) Мнимая часть: {det.imag:.6f}")
```
