# Задача 2.6 - Корни многочлена

---

## 1. Условие

Для многочлена

$$
p(x) = x^{11} + x^{10} + x^8 - 2x^7 - x^5 + x^4 + 2x^2 + 2x + 2
$$

найдите все его 11 корней: $z_1, z_2, \dots, z_{11} \in \mathbb{C}$. В ответе укажите:
1) сумму модулей $|z_1| + \dots + |z_{11}|$;
2) действительную часть корня, у которого эта действительная часть максимальна.

---

## 2. Теоретическая часть

### Нахождение корней многочлена
Для нахождения корней многочлена высокой степени в символьных вычислениях используется метод `nroots` или `solve`. Поскольку многочлен имеет 11-ю степень, аналитическое решение в радикалах может быть громоздким, поэтому рационально использовать численные методы высокой точности, предоставляемые библиотекой `sympy`.

### Комплексные числа и модуль
Корни многочлена с действительными коэффициентами могут быть как действительными, так и комплексными. Комплексные корни всегда появляются сопряжёнными парами.
Модуль комплексного числа $z = a + bi$ вычисляется как $|z| = \sqrt{a^2 + b^2}$.

### План решения
1. Задать многочлен $p(x)$.
2. Найти все его корни (используя `nroots` для численного решения или `solve`).
3. Вычислить сумму модулей всех корней.
4. Найти корень с максимальной действительной частью.

> В Python для точных символьных вычислений удобно использовать `sympy`.

---

## 3. Решение

```python
import sympy as sp
from sympy import init_printing

init_printing()

x = sp.symbols('x')

p = x**11 + x**10 + x**8 - 2*x**7 - x**5 + x**4 + 2*x**2 + 2*x + 2

roots = sp.nroots(p)

sum_abs = sum(abs(r) for r in roots)


max_real_root = max(roots, key=lambda r: sp.re(r))
max_real_part = sp.re(max_real_root)

print("Корни многочлена:")
for i, z in enumerate(roots, start=1):
    print(f"z_{i} = {z}")

print(f"\n 1) Сумма модулей корней: {sum_abs}")
print(f"\n 2) Максимальная действительная часть: {max_real_part}")
print(f"   \n (Соответствующий корень: {max_real_root})")
```
